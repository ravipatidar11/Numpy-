```python
#Numpy--Slicing Operations on 1-D
```

```python
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90])
print("Content of a")
print(a)
print("Dim of a =",a.ndim)
print("Shape of a =",a.shape)
```

```python
#Get 10 20 30 40  from 1D Array
print(a[0:4])
```

```python
#Get 10 20 30 40  from 1D Array
print(a[:4])
```

```python
#Get 10 20 30 40  from 1D Array
print(a[:4:])
```

```python
print(a[::2])
```

```python
print(a[4:1:-1])
```

```python
print(a[::-2])
```

```python
#Numpy--Slicing Operations on 2-D
```

```python
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90,25,35,45,55,77,88,99]).reshape(4,4)
print("Content of ndarray")
print(a)
print("Dim of a=",a.ndim)
print("Shape of a=",a.shape)
```

```python
#Get [60,70,80] from 2D Array
print(a[1:2,1:4])
```

```python
#Get [60,70,80] from 2D Array
print(a[1:2,1:])
```

```python
#Get [60,70,80] from 2D Array with -ve Indices
print(a[-3:-2,-3:])
```

```python
print(a)
```

```python
#Get 60,70,25,35 from 2D Array
print(a[1:3,1:3])
```

```python
#Get 60,70,25,35 from 2D Array with -VE Indices
print(a[-3:-1,-3:-1,])
```

```python
print(a)
```

```python
# Get 10,50,90,55,30,70,35,88
print(a[::,::2])
```

```python
#Get 40,80,45,99
print(a[::,3:])
```

```python
#Get 40,80,45,99
print(a[::,-1:])
```

```python
print(a)
```

```python
# Get 10,55,40,99
print(a[::3,::3])
```
