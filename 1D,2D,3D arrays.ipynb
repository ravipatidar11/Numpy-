```python
import numpy as np
lst=[10,20,30,40,50,60,70,80,90]
a = np.array(lst)
print("Dimention of ndarray =",a.ndim)
print("Content of ndarray =",a)
```

```python
import numpy as np
lst=[10,20,30,40,50,60,70,80,90]
a = np.array(lst)
print("Content of ndarray =",a)
a.shape=(3,3) #reshape the ndarray into 3*3 matrix
print("Dimention of ndarray =",a.ndim)
print(a)
```

```python
import numpy as np
lst=[10,20,30,40,50,60,70,80,90,10,20,30]
a = np.array(lst)
print("Content of ndarray =",a)
a.shape=(2,2,3)
print("Dimention of ndarray =",a.ndim)
print(a)
```
