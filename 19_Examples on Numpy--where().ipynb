```python
#Create an object of ndarray
import numpy as np
a=np.random.randint(100,150,size=10)
print("NDArray Object Data")
print(a)
#Search for those Elements which are Divisible by 5
b=np.where(a%5==0)
print(f"Content of b={b[0]}  Type={type(b[0])}")
```

```python
print(a[b[0]])
```

```python
#Search for those Elements which are Divisible by 5
b=np.extract(a%5==0,a)
print(f"Content of b={b}  Type={type(b)}")
```

```python
#Another Example
arr=np.array([10,15,20,25,30])
idx=np.where(arr>20)
print("Indices =",idx[0])
print("Value =",arr[idx])
```

```python
#Another Example
arr=np.array([2,3,4,5,6,7,8])
idx=np.where(arr%2==0)
print("Indices =",idx[0])
print("Even Elements =",arr[idx])
```

```python
#Another Example
arr=np.array([2,3,4,5,6,7,8])
idx=np.where(arr%2==0,"Even","Odd")
print(idx)
```

```python
nda=np.array([1000,2000,3000,450,550,500,750,5000])
nss=np.where(nda<=1000,nda+nda*50/100,nda)
print(nss)
```

```python
a=np.array([1,2,3,4])
print(a)
nss=np.where( (a < 2) | (a > 3) , 0 , a)
print(nss)
```

```python
a=np.array([1,2,3,4])
print(a)
nss=np.where( (a >= 2) & (a <= 3))
print(a[nss])
```
