```python
#Syntax-1: varname=numpy.random.randint(Value)-----------------> Random Number from 0 to Value-1
```

```python
import numpy.random as r
val = r.randint(5)
print("Random Value = {}  Type = {}".format(val,type(val)))
```

```python
for i in range(1,6):
    val = r.randint(5)
    print("Random Value = {}  Type = {}".format(val,type(val)))
```

```python
#Syntax-2:varname=numpy.random.randint(Low,High)-----> Random Number from Low to High-1
```

```python
val = r.randint(1,50)
print("Random Value = {}  Type = {}".format(val,type(val)))
```

```python
for i in range(1,6):
    val = r.randint(1,50)
    print("Random Value = {}  Type = {}".format(val,type(val)))
```

```python
#Syntax-3:varname=numpy.random.randint(Low,High,size)
```

```python
val = r.randint(1,50,5)
print("Random Value = {}  Type = {}".format(val,type(val)))
```

```python
val = r.randint(1,50,(5,5))
print("Random Value = {}  Type = {}".format(val,type(val)))
```

```python
val = r.randint(1,50,(2,5,5))
print("Random Value = {}  Type = {}".format(val,type(val)))
```
