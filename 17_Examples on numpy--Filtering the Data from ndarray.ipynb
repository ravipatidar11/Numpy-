```python
#Get the Factors of 5 --Normal Python Code
a=[43 ,54, 44, 26, 33, 50, 29, 25, 28, 43, 31, 35]
print(a)
print("----------------------------------------------")
lst=[]
for i in a:
    if(i%5==0):
        lst.append(i)
print("Factors of 5")
print("----------------------------------------------")
print(lst)
```

```python
import numpy.random as r
a=r.randint(25,55,12)
print("Random Elements")
print(a)
#Get all Factors of 5
#Step-1: Prepare the Condition with ndarray object
ba = (a%5==0)
print("------------------------------------------------")
print("Boolean Array={}  Type={}".format(ba,type(ba)))
print("------------------------------------------------")
print("Factors of 5:")
print(a[ba])
```

```python
import numpy.random as r
a=r.randint(25,55,12)
print("Random Elements")
print(a)
#Get all Factors of 5
#Step-1: Prepare the Condition with ndarray object
ba = (a%5==0) & (a%3==0)
print("------------------------------------------------")
print("Boolean Array={}  Type={}".format(ba,type(ba)))
print("------------------------------------------------")
print("Factors of 5 and 3:")
print(a[ba])
```

```python
import numpy.random as r
a=r.randint(25,55,12)
print("Random Elements")
print(a)
#Get all Factors of 5
print("------------------------------------------------")
print("Factors of 5 and 3:")
print(a[(a%5==0) & (a%3==0)])
```
