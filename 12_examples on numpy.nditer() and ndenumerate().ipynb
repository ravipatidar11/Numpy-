```python
#1. By using Looping Statements
```

```python
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90])
print('---------------------------------------------------')
print("1-DArray Data={}  Dim={}".format(a,a.ndim))
print('---------------------------------------------------')
for val in a:
    print("\t{}".format(val))
print('------------------------------------------')
```

```python
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90]).reshape(3,3)
print('---------------------------------------------------')
print("1-DArray Data\n{}  Dim={}".format(a,a.ndim))
print('---------------------------------------------------')
for rowElement in a:
    for val in rowElement:
        print("\t{}".format(val))
print('------------------------------------------')
```

```python
import numpy as np
a=np.array([10,20,30,40,50,60,70,80]).reshape(2,2,2)
print('---------------------------------------------------')
print("1-DArray Data\n{}  Dim={}".format(a,a.ndim))
print('---------------------------------------------------')
for matrix in a:
    for rowElement in matrix:
        for val in rowElement:
            print("\t{}".format(val))
print('------------------------------------------')
```

```python
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90,1,2,3,4,5,6,7]).reshape(2,2,2,2)
print('---------------------------------------------------')
print("1-DArray Data\n{}  Dim={}".format(a,a.ndim))
print('---------------------------------------------------')
for cluster in a:
    for matrix in cluster:
        for rowElement in matrix:
            for val in rowElement:
                print("\t{}".format(val))
print('------------------------------------------')
```

```python
 #By using   numpy.ndenumerate()
```

```python
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90])
print('---------------------------------------------------')
print("1-DArray Data={}  Dim={}".format(a,a.ndim))
print('---------------------------------------------------')
for val in np.ndenumerate(a):
    print("\t{}".format(val))
print('------------------------------------------')
```

```python
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90])
print('---------------------------------------------------')
print("1-DArray Data={}  Dim={}".format(a,a.ndim))
print('---------------------------------------------------')
for val in np.ndenumerate(a):
    print("\tIndex-{}--->Value-{}".format(val[0][0],val[1]))
print('------------------------------------------')
```

```python
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90]).reshape(3,3)
print('---------------------------------------------------')
print("2-DArray Data\n{}  Dim{}".format(a,a.ndim))
print('---------------------------------------------------')
for val in np.ndenumerate(a):
    print("\tIndex--Row-{}, Col-{}--->Value-{}".format(val[0][0],val[0][1],val[1]))
print('------------------------------------------')
```

```python
import numpy as np
a=np.array([10,20,30,40,50,60,70,80]).reshape(2,2,2)
print('---------------------------------------------------')
print("2-DArray Data\n{}  Dim{}".format(a,a.ndim))
print('---------------------------------------------------')
for val in np.ndenumerate(a):
    print("\tIndex--Matrix-{}, Row-{}, Col-{}--->Value-{}".format(val[0][0],val[0][1],val[0][2],val[1]))
print('------------------------------------------')
```

```python
import numpy as np
a=np.arange(2*3*3).reshape(2,3,3)
print('---------------------------------------------------')
print("2-DArray Data\n{}  Dim{}".format(a,a.ndim))
print('---------------------------------------------------')
for val in np.ndenumerate(a):
    print("\tIndex--Matrix-{}, Row-{}, Col-{}--->Value-{}".format(val[0][0],val[0][1],val[0][2],val[1]))
print('------------------------------------------')
```
