```python
import numpy as np
a=np.array([[10,20,30],[40,50,60],[70,80,90]])
print("Matrix-A")
print(a)
print("Dimension of a=",a.ndim)
print("Shape of A=",a.shape)
print("-----------------------------------------")
b=np.array([[15,25],[35,45],[55,65]])
print("Matrix-B")
print(b)
print("Dimension of b=",b.ndim)
print("Shape of B=",b.shape)
print("-----------------------------------------")

c=np.hstack((a,b))
print("Matrix-C")
print(c)
print("Dimension of c=",c.ndim)
print("Shape of C=",c.shape)
print("-----------------------------------------")
```

```python
import numpy as np
a=np.arange(9).reshape(3,3)
print("Matrix-A")
print(a)
print("Dimension of a=",a.ndim)
print("Shape of A=",a.shape)
print("-----------------------------------------")
pd=np.diag(a)
print("Principle Diagonal=",pd)
```
