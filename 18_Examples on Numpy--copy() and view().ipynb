```python
#Implement copy()
#Create ndarray object of ndarray
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90])
print(f"Content of Original Object a={a}  Id={id(a)}")
```

```python
#Copy the Content of ndarray object
b=np.copy(a)
print(f"Content of Copied Object b={b}  Id={id(b)}")
```

```python
#Do the Changes in Original Object and Copied Object
a[0]=1000
b[-1]=2000
```

```python
print(f"Content of Original Object a={a}  Id={id(a)}")
print(f"Content of Copied Object b={b}  Id={id(b)}")
```

```python
#Check the Base for Copied Object
print("Base for Cpoied Object=",b.base)
```

```python
#Check the Base for ndarray Object
print("Base for Original Object=",a.base)
```
