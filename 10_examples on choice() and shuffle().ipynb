```python
#Syntax:   numpy.random.shuffle(ndarray)
```

```python
import numpy as np
a=np.arange(1,6)
print(f"Original Content = {a} ,ID of a ={id(a)}")
#Re-arrange the Elements of ndarray in Itself (inplace)
np.random.shuffle(a)
print(f"Shuffled Content = {a} ,ID of a ={id(a)}")
```

```python
a=np.arange(1,6)
print(f"Original Content = {a} ,ID of a ={id(a)}")
#Re-arrange the Elements of ndarray in Itself (inplace)
print("-"*40)
for i in range(1,6):
    np.random.shuffle(a)
    print(f"{i} Time:Shuffled Content = {a} ,ID of a ={id(a)}")
print("-"*40)
```

```python
a=np.array
print(f"Original Content = {a} ,ID of a ={id(a)}")
#Re-arrange the Elements of ndarray in Itself (inplace)
print("-"*40)
for i in range(1,6):
    np.random.shuffle(a)
    print(f"{i} Time:Shuffled Content = {a} ,ID of a ={id(a)}")
print("-"*40)
```
