```python
#Insert a Row in 2DArray
#Create an object of ndarray 
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90]).reshape(3,3)
print("2ddarray Data")
print(a)
print('---------------------------------------')
```

```python
#Insert row [11,22,33] at 1 Row of 2d
a=np.insert(a,1,[11,22,33],axis=0)
```

```python
print(a)
```

```python
#Insert at 1 column [55,65,75,85] of 2d
a=np.insert(a,1,[55,65,75,85],axis=1)
```

```python
print(a)
```

```python
#Insert row [10,20,30,40] at Last in 2d
a=np.insert(a,len(a),[10,20,30,40],axis=0)
```

```python
print(a)
```

```python
#Insert Column [11,21,31,41,51] at Last in Begining
a=np.insert(a,0,[11,21,31,41,51],axis=1)
```

```python
print(a)
```

```python
#Update all the 3rd Col Values as [21,22,23,24,25]
a[0:1, 3:4]=21
a[1:2, 3:4]=22
a[2:3, 3:4]=23
a[3:4, 3:4]=24
a[4:5, 3:4]=25
```
