```python
#Sort the 1D Array Data
import numpy as np
a=np.random.randint(100,234,size=8)
print("1D Random Data--original ")
print(a)
#sort the Random data
a = np.sort(a)
print("1D Sorted Data in Ascending Order")
print(a)
```

```python
import numpy as np
a=np.random.randint(100,234,size=8)
print("1D Random Data--original ")
print(a)
#sort the Random data
a = np.sort(a)
print("1D Sorted Data in Ascending Order")
print(a)
print("1D Sorted Data in Decsending Order")
a=a[::-1]
print(a)
```

```python
#Sort the 2D Array Data
import numpy as np
a=np.array([15,10,25,11,12,18,29,31,13]).reshape(3,3)
print("2D Random Data--original ")
print(a)
#sort the 2D Array data
a=np.sort(a)
print("2D Sorted Data in Ascending Order")
print(a)
```

```python
#Sort the 2D Array Data
import numpy as np
a=np.array([15,10,25,11,12,18,29,31,13]).reshape(3,3)
print("2D Random Data--original ")
print(a)
#sort the 2D Array data
a=np.sort(a,axis=1)  # Here axis=1 Represents Rows (Default axis=1 -->Rows)
print("2D Sorted Data in Ascending Order")
print(a)
```

```python
#program for Sorting Programmer-Defined Tuples of List Data--Customized Data
dt=[("name","U7"),("age",int),("height",int)]
lst=[("Ramesh",35,175),("Rajesh",40,165),("Rakesh",34,171),("Rossum",60,162),("Raju",35,174),("Ranjeet",40,169),("Ramesh",35,175)]
a=np.array(lst,dtype=dt)
print("Original Data")
for record in a:
    print(record)
print("----------------------------------------------------")   
a=np.sort(a,order="name")
print("Sorted Data Based on Name--Ascending Order")
for record in a:
    print(record)
print("----------------------------------------------------")   
```
