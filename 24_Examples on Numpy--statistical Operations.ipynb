```python
#Create an object of ndarrya Array
import numpy as np
a=np.array([10,2,13,4,14,5,15,18,8])
print("1d Array Elements")
print(a)
```

```python
#Find Highest Element in 1DArray---amax()
#Syntax:     varname=numpy.amax(ndarray,axis)
```

```python
b=np.amax(a)
print("Max Element=",b)
```

```python
#Find Max Element from 2D Array
import numpy as np
a=np.array([10,2,13,4,14,5,15,18,8]).reshape(3,3)
print("2d Array Elements")
print(a)
print("-----------------------------------")
maxv=np.amax(a)
print("Max Element=",maxv)
```

```python
#Find Max Element from 2D Array in Coloumn-wise
import numpy as np
a=np.array([10,2,13,4,14,5,15,18,8]).reshape(3,3)
print("2d Array Elements")
print(a)
print("-----------------------------------")
maxv=np.amax(a)
print("Max Element=",maxv)
colmaxv=np.amax(a,axis=0)
print("Max Col-Wise=",colmaxv)
```

```python
#Find Max Element from 2D Array in Row-wise
import numpy as np
a=np.array([10,2,13,4,14,5,15,18,8]).reshape(3,3)
print("2d Array Elements")
print(a)
print("-----------------------------------")
maxv=np.amax(a)
print("Max Element=",maxv)
rowmaxv=np.amax(a,axis=1)
print("Max Row-Wise=",rowmaxv)
```

```python
#Find Smallest Element in 1DArray---amin()
#Syntax:     varname=numpy.amin(ndarray,axis)
```

```python
#Create an object of ndarrya Array
import numpy as np
a=np.array([10,2,13,4,14,5,15,18,8])
print("1d Array Elements")
print(a)
b=np.amin(a)
print("Min Element=",b)
```

```python
#Find Min Element from 2D Array
import numpy as np
a=np.array([10,2,13,4,14,5,15,18,8]).reshape(3,3)
print("2d Array Elements")
print(a)
print("-----------------------------------")
minv=np.amin(a)
print("Max Element=",minv)
```

```python
#Find Min Element from 2D Array in Column-wise
import numpy as np
a=np.array([10,2,13,4,14,5,15,18,8]).reshape(3,3)
print("2d Array Elements")
print(a)
print("------------------------------------")
minv=np.amin(a)
colminv=np.amin(a,axis=0) # Here axis=0 Represents Cols
print("Min Element=",minv)
print("Column Min Values=",colminv)
print("------------------------------------")
```

```python
#Find Min Element from 1D Array in Row-wise
import numpy as np
a=np.array([10,2,13,4,14,5,15,18,8]).reshape(3,3)
print("2d Array Elements")
print(a)
print("------------------------------------")
minv=np.amin(a)
rowminv=np.amin(a,axis=1) # Here axis=1 Represents Rows
print("Min Element=",minv)
print("Column Min Values=",rowminv)
print("------------------------------------")
```

```python
#Find Min Element from 3D Array in Row-wise
import numpy as np
a=np.array([10,2,13,4,14,5,15,18,8,12,9,10,25,8,4,14]).reshape(2,4,2)
print("3d Array Elements")
print(a)
print("------------------------------------")
maxv=np.amax(a)
print("Max Element=",maxv)
Fcolmaxv=np.amax(a[0],axis=0)
Scolmaxv=np.amax(a[1],axis=0)
print("First Matrix Col Max Vals\n",Fcolmaxv)
print("Second Matrix Col Max Vals\n",Scolmaxv)
```

```python
#Finding the Mean of ndarray Elements--mean()
#Syntax:   varname=numpy.mean(ndarray,axis)
```

```python
#Consider the following ndarray and find mean
import numpy as np
a=np.array([2,3,1,4])
print("1D Array Data")
print(a)
mv=np.mean(a)
print("Mean of ndarray=",mv)
```

```python
#Consider the following 2darray and find mean
import numpy as np
a=np.array([2,3,1,4]).reshape(2,2)
print("2D Array Data")
print(a)
print("-------------------------")
mv=np.mean(a)
print("Mean of ndarray=",mv)
print("-------------------------")
colmv=np.mean(a,axis=0)
print("Col Mean of ndarray=",colmv)
print("-------------------------")
rowmv=np.mean(a,axis=1)
print("Row Mean of ndarray=",rowmv)
```

```python
import numpy as np
a=np.random.randint(10,100,size=(2,3,3))
print("2D Array Data")
print(a)
print("-------------------------")
mv=np.mean(a)
print("Mean of ndarray\n",mv)
print("-------------------------")
colmv=np.mean(a,axis=0)
print("Col Mean of ndarray\n",colmv)
print("-------------------------")
rowmv=np.mean(a,axis=1)
print("Row Mean of ndarray\n",rowmv)
```

```python
#Find the Mean of Opp Principle Diagnoal Elements--
import numpy as np
a=np.random.randint(2,10,size=(3,3))
print("2D Array Data")
print(a)
pd=np.diag(a)
print("Principle Diagnal Elements=",pd)
pdmv=np.mean(pd)
print("Mean of Principle Diagnal Elements=",pdmv)
```

```python
#Find the Mean of Opp. Principle Diagnoal Elements
#Find the Mean of Opp. Principle Diagnoal Elements
a=np.random.randint(2,10,size=(3,3))
print("2D Array Data")
print(a)
b=a[[0,1,2],[2,1,0]]
print("OPP Principle Diagnoal Elements=",b)
mv=np.mean(b)
print("Mean of OPP Principle Diagnoal Elements=",mv)
```

```python
#Find  the Median of ndarray---median()
#Syntax:   vraname=numpy.median(ndarray,axis)
```

```python
a=np.random.randint(12,50,size=7)
print("1D Array Data")
print(a)
#Median  of 1D Array of odd Number of Elements
mdv=np.median(a)
print("Median of 1D Array=",mdv)
```

```python
a=np.random.randint(12,50,size=8)
print("1D Array Data")
print(a)
#Median  of 1D Array of Even Number of Elements
mdv=np.median(a)
print("Median of 1D Array=",mdv)
```

```python
#Find the median of 2d Array
a=np.random.randint(12,50,size=(3,3))
print("2D Array Data")
print(a)
print("------------------------------------")
#Median of 2D
mdv=np.median(a)
colmdv=np.median(a,axis=0)
rowmdv=np.median(a,axis=1)
print("Median=",mdv)
print("Col Median=",colmdv)
print("Row Median=",rowmdv)
```

```python
#Find the Variance of ndarray--most Imp
#Syntax1:-      varname=numpy.var(ndarrayobject)
#Syntax2:-      varname=numpy.var(ndarrayobject,axis=0)
#Syntax3:-      varname=numpy.var(ndarrayobject,axis=1)
```

```python
#Find the Variance of ndarray--most Imp
import numpy as np
a=np.array([2,3,1,4])
print("1D Array Data")
print(a)
v=np.var(a)
print("Variance=",v)
```

```python
#Find the Variance of ndarray--most Imp
import numpy as np
a=np.array([2,3,1,4]).reshape(2,2)
print("---------------------------------")
print("2D Array Data")
print(a)
print("---------------------------------")
v=np.var(a)
cv=np.var(a,axis=0)
rv=np.var(a,axis=1)
print("---------------------------------")
print("Variance=",v)
print("Col Variance=",cv)
print("Row Variance=",rv)
```
