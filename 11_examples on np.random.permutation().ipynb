```python
#Re-arrange the Elements of ndarray by using permutation()
import numpy.random as r
a=r.randint(10,21,size=7)
print("Original Content Before permutation ={}".format(a))
print("---------------------------------------------------------------")
ra=r.permutation(a)
print("Original Content After permutation ={}".format(a))
print("---------------------------------------------------------------")
print("Re-Arranged Elements of ra ={}".format(ra))
```

```python
#Re-arrange the Elements of ndarray by using permutation()
import numpy.random as r
a=r.randint(10,21,size=7)
print("Original Content Before permutation ={}".format(a))
print("-------------------------------------------------------------------------------")
for i in range(1,len(a)+1):
    ra=r.permutation(a)
    print("{}:Time, Re-Arranged Elements of ra ={} ID ={}".format(i,ra,id(ra)))
print("-------------------------------------------------------------------------------")
```

```python
#Re-arrange the Elements of ndarray by using permutation()
import numpy.random as r
import numpy as np
a=np.array(list("MISSISSIPPI"))
print("Original Content Before permutation ={}".format(a))
print("-------------------------------------------------------------------------------")
for i in range(1,6):
    ra=r.permutation(a)
    print("Re-Arranged Elements of ra ={}".format("".join(ra)))
print("-------------------------------------------------------------------------------")
```
