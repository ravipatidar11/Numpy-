```python
#Code for Building  a Matrix with dim of 2 with All entries as 0
import numpy as np
lst=[0,0,0,0]
a=np.array(lst).reshape(2,2)
print("Content of a\n{}".format(a))
print("Dimension of a = {}".format(a.ndim))
```

```python
#Create an object of ndarray with 6 Zeros 
a=np.zeros(6)
print("Content of a\n{}".format(a))
print("Dimension of a = {}".format(a.ndim))
```

```python
#Create an object of ndarray with 6 Zeros 
a=np.zeros(6,int)
print("Content of a\n{}".format(a))
print("Dimension of a = {}".format(a.ndim))
```

```python
#Create an object of ndarray with 9 Zeros with 2-D
a=np.zeros((3,3),dtype=int)
print("Content of a\n{}".format(a))
print("Dimension of a = {}".format(a.ndim))
```

```python
#Create an object of ndarray with 16 Zeros with 4-D
a=np.zeros((2,2,2,2),dtype=int)
print("Content of a\n{}".format(a))
print("Dimension of a = {}".format(a.ndim))
```

```python
#Create an object of ndarray with 6 Ones
a=np.ones(6,int)
print("Content of a\n{}".format(a))
print("Dimension of a = {}".format(a.ndim))
```

```python
#Create an object of ndarray with 16 Ones
a=np.ones((2,2,2,2),int)
print("Content of a\n{}".format(a))
print("Dimension of a = {}".format(a.ndim))
```

```python
#Create an object of ndarray with 6 Entries as 3
a=np.full(6,3)
print("Content of a\n{}".format(a))
print("Dimension of a = {}".format(a.ndim))
```

```python
#Create an object of ndarray with 12 Entries as 4
a=np.full((3,4),4)
print("Content of a\n{}".format(a))
print("Dimension of a = {}".format(a.ndim))
```

```python
#Create an object of ndarray with 12 Entries as 4
a=np.full(fill_value=4,shape=(3,4))
print("Content of a\n{}".format(a))
print("Dimension of a = {}".format(a.ndim))
```

```python
#Syntax:  varname=numpy.linspace(start,stop,num=50,endpoint=True,retstep=False,dtype)
```

```python
#Syntax:  varname=numpy.linspace(start,stop,num=50,endpoint=True,retstep=False,dtype)
a=np.linspace(0,10,num=20)
print("Content of a\n{}".format(a))
print("Dimension of a = {}".format(a.ndim))
```

```python
#Syntax:  varname=numpy.linspace(start,stop,num=50,endpoint=True,retstep=False,dtype)
a=np.linspace(0,10,num=10,endpoint=False)
print("Content of a\n{}".format(a))
print("Dimension of a = {}".format(a.ndim))
```

```python
#Syntax:  varname=numpy.linspace(start,stop,num=50,endpoint=True,retstep=False,dtype)
a=np.linspace(0,10,num=10,endpoint=False,retstep=True)
print("Content of a\n{}".format(a[0]))
print("-------------------------------------------------------------------------")
print("Return Step of a =",a[1])
```

```python
#Syntax:  varname=numpy.linspace(start,stop,num=50,endpoint=True,retstep=False,dtype)
a=np.linspace(0,10,num=40,endpoint=False,retstep=True)
print("Content of a\n{}".format(a[0]))
print("-------------------------------------------------------------------------")
print("Return Step of a =",a[1])
```

```python
#=>Syntax:  varname=numpy.identity(N,dtype)
a=np.identity(3)
print("Content of a\n{}".format(a))
print("Dimension of a = {}".format(a.ndim))
```
