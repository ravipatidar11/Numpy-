```python
#varname=numpy.arange(Value)
import numpy as np
a=np.arange(16).reshape(2,2,2,2)
print("content of a\n{}  Type={}".format(a,type(a)))
print("Dimension ndarray ={}".format(a.ndim))  #  4 
print("Shape of ndarray={}".format(a.shape))   #  (2,2,2,2)
print("Size of ndarray={}".format(a.size)) #  16
```
