```python
#Create 3D Array with Random Integer Values with shape(3,3,3)
```

```python
import numpy.random as r
a=r.randint(10,20,size=(3,3,3))
print("3D Array")
print(a)
```

```python
#Get First Two Matrices from 3D
print(a[0:2])
```

```python
#get First and Last Matrix
print(a[::2])
```

```python
print(a)
```

```python
#Get All 0th row of all Columns of all Matrices
print(a[::,0:1,::])
```

```python
#Get All 0th Col of all Rows of all Matrices
print(a[::,::,:1])
```

```python
print(a)
```

```python
#Get First and Last Matrices 2nd Row Elements of Correcposnding Matrices
print(a[::2,1:2,::])
```

```python
# Get last matrix last row last last two elements [13,14]
print(a[2:,2:,1:3])
```
