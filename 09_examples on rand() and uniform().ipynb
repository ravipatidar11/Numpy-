```python
#5. uniform()  :  Uniform Distribution
#------------------------------------------------------------------------
#This Function Generates random sample(s) between Two Intervals(Low,High) According Programmer Choice.
#The Default random sample is low is 0.0 and high is 1.0
#Syntax-1:      varname=numpy.random.uniform()
#Syntax-2:	   varname=numpy.random.uniform(Low,High)	
#Syntax-3:	   varname=numpy.random.uniform(Low,High,size)
```

```python
from numpy import random as r
val=r.uniform()
print("Random Value = {}  Type = {}".format(val,type(val)))
```

```python
val=r.uniform()
print("Random Value = {}  Type = {}".format(val,type(val)))
```

```python
val=r.uniform(3)
print("Random Value = {}  Type = {}".format(val,type(val)))
```

```python
val=r.uniform(3,5)
print("Random Value = {}  Type = {}".format(val,type(val)))
```

```python
val=r.uniform(10,22)
print("Random Value = {}  Type = {}".format(val,type(val)))
```

```python
val=r.uniform(10,22,5)
print("Random Value = {}  Type = {}".format(val,type(val)))
```

```python
val=r.uniform(19.)
print("Random Value = {}  Type = {}".format(val,type(val)))
```
