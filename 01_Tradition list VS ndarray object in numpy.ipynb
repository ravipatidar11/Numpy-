```python
#Create an Object of List
print("-"*40)
import sys
lst=[10,20,30,40,50,60,70,15,23,56,78,90,34,66,89,123,45,78]
print("Content of Lst=",lst)
print("Type of lst=",type(lst))
print("Memory of Traditional Python List=",sys.getsizeof(lst))
print("-"*40)
```

```python
#Convert Traditional Python List into ndarray object
print("-"*50)
import numpy as np
lst=[10,20,30,40,50,60,70,15,23,56,78,90,34,66,89,123,45,78]
print("Content of Lst=",lst)
a=np.array(lst)
print("Content of a={}  Type={}".format(a,type(a)))
print("Memory Space of ndarray Object a=",sys.getsizeof(a))
print("-"*50)
```
