```python
#Syntax:varname=numpy.random.randn()----->Gives Random Floating Point Value 
```

```python
import numpy.random as r
val=r.randn()
print("Random floating value = {}  Type = {}".format(val,type(val)))
```

```python
val=r.randn()
print("Random floating value = {}  Type = {}".format(val,type(val)))
```

```python
for i in range(1,6):
    val=r.randn()
    print("Random floating value = {}  Type = {}".format(val,type(val)))
    
```

```python
#Syntax:varname=numpy.random.randn(Value)
#Gives  Number of Random Floating Point Values Depends on 'Value'-- +VE OR -VE---1D
```

```python
val=r.randn(6)
print("Random floating value = {}  Type = {}".format(val,type(val)))
```

```python
#Syntax: varname=numpy.random.randn(Rows,Cols)
#Gives  Number of Random Floating Point Values Depends on 'Rows x Cols'-- +VE OR -VE---2D
```

```python
val=r.randn(6,6)
print("Random floating value = {}  Type = {}".format(val,type(val)))
```

```python
val=r.randn(2,3,3)
print("Random floating value = {}  Type = {}".format(val,type(val)))
```

```python
val=r.randn(2,3,3)
print("Random floating value = {}  Type = {}".format(val,type(val)))
```

```python
#Syntax:    varname=numpy.random.normal(loc=0,scale=1,size=None)
```

```python
val=r.normal()
print("Random floating value = {}  Type = {}".format(val,type(val)))
```

```python
for i in range(1,6):
    val=r.normal()
    print("Random floating value = {}  Type = {}".format(val,type(val)))
```
