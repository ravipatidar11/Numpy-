```python
#To delete the Element from 1D Array-we use delete()
#Syntax-1:    numpy.delete(1dArray, Index)
#Syntax-2:    numpy.delete(1dArray, [Index1,Index2,...,Index-n] )
```

```python
#To delete the Elements(Rows/Columns) from 2D Array-we use delete()
#Syntax-1: numpy.delete(2dArray, Index,axis=0)  # deletes Row from 2D Based Row Index
#Syntax-2:numpy.delete(2dArray,[Index1,Index2,...,Index-n],axis=0 )  
# deletes Rows from 2D Based Row Indices
#Syntax-3: numpy.delete(2dArray, Index,axis=1)  # deletes Column from 2D Based Column Index
#Syntax-2:numpy.delete(2dArray,[Index1,Index2,...,Index-n],axis=1 )  
# deletes Columns from 2D Based Column Indices
```

```python
#Create 1D Array 
import numpy as np
a=np.array([10,20,30,50,60,70,80,90])
print("Original 1D Array Data")
print(a)
#Delete  30 from 1D
a=np.delete(a,2)
print("1D Array Data after delete")
print(a)
```

```python
#Create 1D Array 
import numpy as np
a=np.array([10,20,30,50,60,70,80,90])
print("Original 1D Array Data")
print(a) # [10 20 30 50 60 70 80 90]
#Delete  30 , 70 and 80 from 1D
a=np.delete(a,[2,-3,-2])
print("1D Array Data after delete")
print(a)
```

```python
#Create 2D Array 
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90]).reshape(3,3)
print("Original 2D Array Data")
print(a) 
#Delete 1 Row from 2D 
a=np.delete(a,1,axis=0)
print("2D Array Data After Row Delete")
print(a) 
```

```python
#Create 2D Array 
import numpy as np
a=np.arange(5*5).reshape(5,5)
print("Original 2D Array Data")
print(a) 
#Delete 1 Column from 2D 
a=np.delete(a,1,axis=1)
print("2D Array Data After Column Delete")
print(a)     
```

```python
#Create 2D Array 
import numpy as np
a=np.arange(5*5).reshape(5,5)
print("Original 2D Array Data")
print(a) 
#Delete 0th , last and last but one Column from 2D 
a=np.delete(a,[0,-2,-1],axis=1)
print("2D Array Data After Column Delete")
print(a) 
```
