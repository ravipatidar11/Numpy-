```python
#Creating an Object of ndarray by using  array()
#int
import numpy as np
a=10
a1=np.array(a)
print("Content of ndarray = {} type of ndarray = {}".format(a1,type(a1)))
print("Dimension of ndarray = {}".format(a1.ndim)) # 0
print("Shape of ndarray = {}".format(a1.shape)) # ()
print("Size of ndarray = {}".format(a1.size)) # 1
```

```python
#Creating an Object of ndarray by using  array()
#list
import numpy as np
lst=[[10,20],[30,40],[50,60],[70,80]]
a=np.array(lst,float)
print("Content of ndarray\n{}  Type={}".format(a,type(a)))
print("Dimension ndarray ={}".format(a.ndim))  #  2
print("Shape of ndarray={}".format(a.shape))   #  (4,2)
print("Size of ndarray={}".format(a.size)) # 8
print("Data type Elements of ndarray={}".format(a.dtype)) # float64
```
