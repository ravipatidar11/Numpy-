```python
#Create Two Matrices with 2 D of Shape (2,2)
import numpy as np
a=np.array([10,20,30,40]).reshape(2,2)
b=np.array([1,2,3,4]).reshape(2,2)
print("Matrix-a")
print(a)
print("----------------------------------")
print("Matrix-b")
print(b)
```

```python
#Perform Addition of Matrices ---add()
#Syntax:   varname=numpy.add(ndarray1,mdarray2)
import numpy as np
a=np.array([10,20,30,40]).reshape(2,2)
b=np.array([1,2,3,4]).reshape(2,2)
print("Matrix-a")
print(a)
print("----------------------------------")
print("Matrix-b")
print(b)
print("----------------------------------")
c=np.add(a,b)
print("Addition of a and b")
print(c)
```

```python
#Perform dot Product of Matrices ---dot() or matmul()
#Rule:  Number of Columns in ndarray1 Must be equal to Number of rows in ndarray2
#    Otherwise dot() or matmul() is Not Possible
#Syntax:   varname=numpy.dot(ndarray1,ndarray2)
import numpy as np
a=np.array([10,20,30,40]).reshape(2,2)
b=np.array([1,2,3,4]).reshape(2,2)
print("Matrix-a")
print(a)
print(a,a.shape)
print("----------------------------------")
print("Matrix-b")
print(b)
print(b,b.shape)
print("----------------------------------")
c=np.dot(a,b)
print("Multiplication of a and b")
print(c)
print(c,c.shape)
```
