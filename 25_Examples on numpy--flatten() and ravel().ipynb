```python
#Create an object of ndarray 
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90]).reshape(3,3)
print("Matrix-A")
print(a)
print("Dim of a=",a.ndim)
print("shape of a=",a.shape)
print('---------------------------------------')
#flatten the ndarray Object
b=a.flatten()
print("Content of b")
print(b)
print("Dim of b=",b.ndim)
print("shape of b=",b.shape)
print('---------------------------------------')
```

```python
#Flatten the ndarray data by using ravel()
#Create an object of ndarray 
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90]).reshape(3,3)
print("Matrix-A")
print(a)
print("Dim of a=",a.ndim)
print("shape of a=",a.shape)
print('---------------------------------------')
#flatten the ndarray Object--ravel()
b=a.ravel()
print("Content of b")
print(b)
print("Dim of b=",b.ndim)
print("shape of b=",b.shape)
print('---------------------------------------')
```

```python
print("ID of Original ndarray a=",id(a))
print("ID of Ravelled ndarray b=",id(b))
```

```python
#Do the changes on Both ndarray Object
a[0][0]=11
```

```python
print(a)
print("----------------------")
print(b)
```

```python
b[1]=155
```

```python
print(a)
print("----------------------")
print(b)
```

```python
#Update 0th Row of ndarray a
a[0:1,:]=[100,200,300]
```

```python
print(a)
```

```python
print(b)
```

```python
#Check for Base
print("Base of Ravelled Array b=",b.base)
```
